```python
class Node:
  def __init__(self, data):
    self.data = data #menyimpan nilai
    self.left = None #tangan kiri untuk memegang anak kiri
    self.right = None #tangan kanan untuk memegang anak kanan

root = Node(1) #angka 1 menjadi akar paling atas
root.left = Node(2) #anak kiri dari 1
root.right = Node(3) #anak kanan dari 1
root.left.left = Node(4) #anak kiri dari 2
root.left.right = Node(5) #anak kanan dari 2

def inorder(node):
  if node:
    inorder(node.left) #pergi ke paling kiri
    print(node.data, end=' ') #cetak angka tersebut
    inorder(node.right) #pergi ke kanan

print("Hasil inorder traversal dari binary tree:")
inorder(root)
```

```python
class Node:
  def __init__(self, data):
    self.data = data
    self.left = None
    self.right = None

class BinarySearchTree:
  def __init__(self):
    self.root = None
  def insert(self, root, data):
    if root is None: #jika kosong, taruh node baru di sini
      return Node(data)
    if data < root.data:
      root.left = self.insert(root.left, data) #jika lebih kecil, belok ke kiri
    else:
      root.right = self.insert(root.right, data) #jika lebih besar, belok ke kanan
    return root

bst = BinarySearchTree()
root = None
data_list = [50, 30, 70, 20, 40, 60, 80]
for value in data_list:
  root = bst.insert(root, value)
```

```python
def inorder(node):
  if node:
    inorder(node.left)
    print(node.data, end=' ')
    inorder(node.right)

def preorder(node):
  if node:
    print(node.data, end=' ')
    preorder(node.left)
    preorder(node.right)

def postorder(node):
  if node:
    postorder(node.left)
    postorder(node.right)
    print(node.data, end=' ')

print("Inorder Traversal:")
inorder(root)
print("\nPreorder Traversal:")
preorder(root)
print("\nPostorder Traversal:")
postorder(root)
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

class BinarySearchTree:
    def __init__(self):
        self.root = None

    # Fungsi otomatis menyisipkan angka sesuai aturan BST
    def insert(self, root, data):
        if root is None:
            return Node(data)
        if data < root.data:
            root.left = self.insert(root.left, data)
        else:
            root.right = self.insert(root.right, data)
        return root

    # Fungsi mencari angka di dalam tree
    def search(self, node, key):
        if node is None or node.data == key:
            return node
        if key < node.data:
            return self.search(node.left, key)
        return self.search(node.right, key)

# Fungsi Traversal
def inorder(node):
    if node:
        inorder(node.left)
        print(node.data, end=' ')
        inorder(node.right)

def preorder(node):
    if node:
        print(node.data, end=' ')
        preorder(node.left)
        preorder(node.right)

def postorder(node):
    if node:
        postorder(node.left)
        postorder(node.right)
        print(node.data, end=' ')

#eksekusi program

bst = BinarySearchTree()
# Daftar 7 angka sesuai ketentuan (NPM, DD, MM, ASCII M, ASCII a, Bebas1, Bebas2)
data_list = [38, 17, 5, 77, 97, 10, 90]

# Memasukkan data ke dalam BST
root = None
for value in data_list:
    root = bst.insert(root, value)

print("=== HASIL TRAVERSAL ===")
print("Inorder   :", end=" ")
inorder(root)
print("\nPreorder  :", end=" ")
preorder(root)
print("\nPostorder :", end=" ")
postorder(root)

print("\n\n=== FITUR PENCARIAN ===")

# 1. Cari dua digit terakhir NPM (38)
key1 = 38
result1 = bst.search(root, key1)
print(f"Pencarian {key1}: {'Ditemukan' if result1 else 'Tidak Ditemukan'}")

# 2. Cari angka yang tidak ada di list (misal: 100)
key2 = 100
result2 = bst.search(root, key2)
print(f"Pencarian {key2}: {'Ditemukan' if result2 else 'Tidak Ditemukan'}")
```

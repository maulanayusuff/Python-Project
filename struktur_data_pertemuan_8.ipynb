```python
graph = {
    'A': ['B', 'C'], #node a bertetangga dengan node b dan c
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

def print_graph(graf):
  for node in graf:
    print(f"{node} -> {graf[node]}") #menampilkan nama node dan list tetangganya

print_graph(graph)
```

```python
from collections import deque

def bfs(graph, start):
  visited = set() #tempat menyimpan node yang sudah dikunjungi
  queue = deque([start])

  while queue:
    vertex = queue.popleft() #ambil node paling depan dari antrian
    if vertex not in visited:
      print(vertex, end=" ") #menampilkan nama node
      visited.add(vertex) #memasukan node ke dalam daftar visited
      queue.extend([neighbor for neighbor in graph[vertex] if neighbor not in visited])

bfs(graph, 'A') #menjalankan fungsi bfs dimulai dari node a
```

```python
def dfs(graph, start, visited=None):
  if visited is None: #jika visited kosong
    visited = set() #buat wadah set kosong untuk mencatat node yang sudah dikunjungi

  if start not in visited:
    print(start, end=" ") #menampilkan nama node
    visited.add(start) #memasukan node ke dalam daftar visited
    for neighbor in graph[start]:
      dfs(graph, neighbor, visited)

dfs(graph, 'A')
```

```python
def count_neighbors(graph):
  result = {}
  for node in graph:
    result[node] = len(graph[node])
  return result

print(count_neighbors(graph))
```

```python
#tugas
#1
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'C', 'D', 'E'],
    'C': ['A', 'B', 'E'],
    'D': ['B', 'E'],
    'E': ['B', 'C']
}

def print_graph(graf):
  for node in graf:
    print(f"{node} -> {graf[node]}")

print_graph(graph)
```

```python
#2
from collections import deque

def bfs(graph, start):
  visited = set()
  queue = deque([start])

  while queue:
    vertex = queue.popleft()
    if vertex not in visited:
      print(vertex, end=" ")
      visited.add(vertex)
      queue.extend([neighbor for neighbor in graph[vertex] if neighbor not in visited])

def dfs(graph, start, visited=None):
  if visited is None:
    visited = set()

  if start not in visited:
    print(start, end=" ")
    visited.add(start)
    for neighbor in graph[start]:
      dfs(graph, neighbor, visited)


bfs(graph, 'A')
print("\n")
dfs(graph, 'A')
```

```python
#3
def bfs_to_list(graph, start):
    visited = set()
    queue = deque([start])
    traversal_order = []

    while queue:
        vertex = queue.popleft()
        if vertex not in visited:
            traversal_order.append(vertex)
            visited.add(vertex)
            queue.extend([neighbor for neighbor in graph[vertex] if neighbor not in visited])
    return traversal_order

bfs_to_list(graph, 'A')
```

```python
#4
def find_path(graph, start, end, path=None):
    if path is None:
        path = []

    path = path + [start]

    if start == end:
        return path

    if start not in graph:
        return None

    for neighbor in graph[start]:
        if neighbor not in path:
            newpath = find_path(graph, neighbor, end, path)
            if newpath:
                return newpath

    return None

jalur_ketemu = find_path(graph, 'A', 'E')
print("Jalur dari A ke E:", jalur_ketemu)
```

```python
#5
def dfs_to_list(graph, start, visited=None):
    if visited is None:
        visited = set()
    if start not in visited:
        visited.add(start)
        for neighbor in graph[start]:
            dfs_to_list(graph, neighbor, visited)
    return visited

def is_connected(graph):
    if not graph:
        return True
    start_node = list(graph.keys())[0]
    simpul_terjangkau = dfs_to_list(graph, start_node)
    return len(simpul_terjangkau) == len(graph)


is_connected(graph)
```

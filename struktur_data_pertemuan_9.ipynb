```python
class SimpleHashTable:
  def __init__(self, size): #fungsi untuk membuat objek baru dengan menentukan ukuran tabel
    self.size= size
    self.table= [None] * size

  def hash_function(self, key): #fungsi untuk mengubah kata kunci (key) menjadi nomor laci/indeks
    return hash(key) % self.size

  def insert(self, key, value): #fungsi untuk memasukkan data baru
    index= self.hash_function(key)
    self.table[index]= value

  def search(self, key): #fungsi untuk mencari nilai data berdasarkan kata kuncinya (key)
    index= self.hash_function(key)
    return self.table[index]

ht= SimpleHashTable(10) #membuat objek tabel hash baru bernama 'ht' dengan total 10 laci kosong
ht.insert("apel", 100) #memasukkan kata "apel" dengan nilai 100 ke dalam tabel
ht.insert("pisang", 200) #memasukkan kata "pisang" dengan nilai 200 ke dalam tabel

print("nilai untuk apel:", ht.search("apel")) #mencari dan mencetak nilai untuk kata "apel"
print("nilai untuk pisang:", ht.search("pisang")) #mencari dan mencetak nilai untuk kata "pisang"
```

```python
class LinearProbingHashTable:
  def __init__(self, size):
    self.size= size
    self.table= [None] * size

  def hash_function(self, key):
    return hash(key) % self.size

  def insert(self, key, value):
    index= self.hash_function(key)
    original_index= index

    while self.table[index] is not None:
      if self.table[index][0]== key:
        break
      index= (index + 1) % self.size
      if index== original_index:
        raise Exception ("hash table penuh")

    self.table[index]= (key, value)

  def search(self, key):
    index= self.hash_function(key)
    original_index= index

    while self.table[index] is not None:
      if self.table[index][0]== key:
        return self.table[index][1]
      index= (index + 1) % self.size
      if index== original_index:
        break
    return None

ht= LinearProbingHashTable(5)
ht.insert("apel", 100)
ht.insert("pisang", 200)
ht.insert("melon", 300)

print("nilai untuk apel:", ht.search("apel"))
print("nilai untuk pisang:", ht.search("pisang"))
print("nilai untuk melon:", ht.search("melon"))
```

```python
class ChainingHashTable:
  def __init__(self, size):
    self.size= size
    self.table= [[] for _ in range(size)]

  def hash_function(self, key):
    return hash(key) % self.size

  def insert(self, key, value):
    index= self.hash_function(key)
    for i, (k, v) in enumerate(self.table[index]):
      if k== key:
        self.table[index][i]= (key, value)
        return
    self.table[index].append((key, value))

  def search(self, key):
    index= self.hash_function(key)
    for k, v in self.table[index]:
      if k== key:
        return v
    return None

ht= ChainingHashTable(5)
ht.insert("apel", 100)
ht.insert("pisang", 200)
ht.insert("melon", 300)

print("nilai untuk apel:", ht.search("apel"))
print("nilai untuk pisang:", ht.search("pisang"))
print("nilai untuk melon:", ht.search("melon"))
```

```python
class ChainingHashTable:
  def __init__(self, size):
    self.size= size
    self.table= [[] for _ in range(size)]

  def hash_function(self, key):
    return sum(ord(char) for char in key) % self.size

  def insert(self, key, value):
    index= self.hash_function(key)
    for i, (k, v) in enumerate(self.table[index]):
      if k== key:
        self.table[index][i]= (key, value)
        return
    self.table[index].append((key, value))

  def search(self, key):
    index= self.hash_function(key)
    for k, v in self.table[index]:
      if k== key:
        return v
    return None

#a. Gunakan ukuran tabel 3
ht= ChainingHashTable(3)

#b. Masukkan key-value berikut: "apel": 100, "melon": 200, "lemon": 300, "pisang": 400

ht.insert("apel", 100)
ht.insert("melon", 200)
ht.insert("lemon", 300)
ht.insert("pisang", 400)

#c. Cetak isi tabel (self.table) dan tampilkan semua data yang disimpan di setiap slot
print(ht.table)
```
